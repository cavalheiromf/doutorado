```python
import pandas as pd
import glob
```

```python
oscasID = pd.read_csv("//home/mfcaval/github/doutorado/chapter2_osca/01_gene_identification/results/oscasIDs.txt", sep="\t", header=None)
oscas_dict = dict(zip(oscasID[0], oscasID[1]))
oscas_dict
```

```python
myFiles = glob.glob("/home/mfcaval/github/doutorado/chapter2_osca/01_gene_identification/results/interproscan_raw/*.tsv")

dd = []

for file in myFiles:
    df = pd.read_csv(file, sep="\t", header=None)
    dd.append(df)

df = pd.concat(dd, ignore_index=True)
```

```python
df[0] = df[0].replace(oscas_dict)
sl = df[df[3] == "Pfam"].copy()
```

```python
sl[sl[0].str.startswith("Vt")].to_csv("/home/mfcaval/github/doutorado/chapter2_osca/01_gene_identification/results/interproscan_raw/pfam_Vt.tsv", sep="\t", index=False, header=False)
sl[sl[0].str.startswith("Vp")].to_csv("/home/mfcaval/github/doutorado/chapter2_osca/01_gene_identification/results/interproscan_raw/pfam_Vp.tsv", sep="\t", index=False, header=False)
```
